```python
def compare_values(new_dict, old_dict, path=""):
    old_dict = old_dict if isinstance(old_dict, dict) else {}
    new_dict = new_dict if isinstance(new_dict, dict) else {}

    all_keys = new_dict.keys() | old_dict.keys()

    for key in all_keys:
        new_value = new_dict.get(key, "")
        old_value = old_dict.get(key, "")

        if isinstance(new_value, dict) or isinstance(old_value, dict):
            compare_values(new_value, old_value, path=f"{path}{key}.")
            continue

        if new_value != old_value:
            print(f"{path + key:<40} {str(new_value):<40}{old_value}")

compare_values(new_plan, old_plan)
```
