```python
import pandas as pd

from rockyelevate.wrapper import Session
from rockyelevate.utils import response_to_dataframe
```

```python
elv = Session(server="UAT", multithread=True, max_threads=40)
```

```python
orgs = elv.get_orgs_by_id(children=False)
```

```python
org_df = response_to_dataframe(orgs)
```

```python
my_org = org_df[org_df['external_identifier'] == "RMRSNOW"]
```

```python
my_org_id = my_org.iloc[0]['id']
```

```python
my_org_id
```

```python
plans = elv.get_plans_by_org(oids=[int(i) for i in [my_org_id]])
```

```python
plans_df = response_to_dataframe(plans)
```

```python
plans_df['plan_year.valid_from'] = pd.to_datetime(plans_df['plan_year.valid_from'])
```

```python
plans_df = plans_df.sort_values("plan_year.valid_from")
```

```python
plans_df['plan_year.valid_from'].value_counts()
```
