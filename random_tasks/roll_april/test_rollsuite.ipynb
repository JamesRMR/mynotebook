```python
elv_update_responses, update_exceptions = roll_suite.send_update_bodies(update_bodies)
```

```python
for request in elv_update_responses:
    for plan_id, response in request.items():
        keys = list(response.keys())

        if keys[0] != "id":
            print(f"error updating plan {plan_id}")
```
