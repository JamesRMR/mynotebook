```python
def get_file_history(path: str = None):
    if not path:
        return []

    history_res = rfc.get_history(path)

    if history_res.has_next_page:
        print("WARNING: file history has another page")

    return [h.__dict__ for h in history_res.data]
```

```python
path = "/Clients/Krayden Inc-RMRKRA/File Feeds/Completed FLEX/ALL ZEROS - Krayden Inc RMRKRA_Contribution_20260910.csv"

history = get_file_history(path)
```
