```python
import json

from rockyclickup.wrapper import Session
```

```python
rcu = Session()
```

```python
with open("task_ids_to_delete.json", "r") as f:
    task_ids_to_delete = json.load(f)
```

```python
task_ids_to_delete
```

```python
responses = []

for task_id in task_ids_to_delete:
    try:
        res = rcu.delete(task_id)
    except json.JSONDecodeError as e:
        res = e
    responses.append({
        "task_id": task_id,
        "response": res
    })
```
