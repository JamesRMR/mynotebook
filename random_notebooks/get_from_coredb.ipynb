```python
from rockydb.connection import CoreDB
```

```python
with CoreDB() as cdb:
    all_db_entities = cdb.get_all_from_table(table="entity")
```

```python
autocomplete_map = {}

for item in all_db_entities:
    if not all([item.elevate_id, item.employer_id, item.name_operational]):
        continue

    autocomplete_map.update(
        {
            item.employer_id: item.elevate_id,
            item.name_operational: item.elevate_id
        }
    )
```

```python
autocomplete_map = dict(sorted(autocomplete_map.items()))
```
