```python
import os, sys

import pandas as pd
import datetime as dt

current_dir = os.getcwd()
parent_dir = os.path.dirname(f"{"\\".join(current_dir.split("\\")[:-1])}")
sys.path.append(parent_dir)

from utils.merger import Merger
```

```python
# open pickles of all clients, organizations, and plans
all_orgs_path = "./cache/all_orgs.pkl"
all_clients_path = "./cache/all_clients.pkl"
all_elevate_plans_path = "./cache/all_elv_plans.pkl"
all_clickup_plans_path = "./cache/all_plans.pkl"


orgs_df = pd.read_pickle(all_orgs_path)
print(len(orgs_df))
clients_df = pd.read_pickle(all_clients_path)
print(len(clients_df))
elevate_plans_df = pd.read_pickle(all_elevate_plans_path)
print(len(elevate_plans_df))
clickup_plans_df = pd.read_pickle(all_clickup_plans_path)
print(len(clickup_plans_df))
```

```python
# merge them together

merger = Merger(
    organization_df=orgs_df,
    elv_plan_df=elevate_plans_df,
    client_df=clients_df,
    cu_plan_df=clickup_plans_df
)

merge_df = merger.merge_all()

merge_df.to_pickle(f"{dt.datetime.now().strftime("%y%m%d")}_merge_df.pkl")
```
