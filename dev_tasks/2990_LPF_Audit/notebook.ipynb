```python
import pandas as pd

from rockyclickup.wrapper import Session
from rockyclickup.models import FSA
from rockyclickup.utils import response_to_dataframe
```

```python
clickup = Session()
```

```python
all_fsas = clickup.get_full_list(model=FSA)

fsa_df = response_to_dataframe(all_fsas)
```

```python
for col in ['date_plan_start', 'date_plan_end']:
    fsa_df[col] = pd.to_datetime(fsa_df[col])
```

```python
fsa_df['client_id'] = fsa_df['client_fsa'].apply(lambda x: x[0] if (isinstance(x, list) and len(x) > 0) else None)
```

```python
lpf_map = {}

for client_id, client_df in fsa_df.groupby("client_id"):

    sorted_df = client_df.sort_values("date_plan_start")

    oldest_lpf_value = bool(sorted_df.iloc[0].get("lpf"))

    lpf_map[client_id] = oldest_lpf_value
```

```python
fsa_df['is_lpf'] = fsa_df['client_id'].map(lpf_map)
fsa_df['is_lpf'].fillna(False, inplace=True)

plans_to_correct = fsa_df[
    (fsa_df['is_lpf'] != fsa_df['lpf']) &
    (fsa_df['client_id'].notna())
]

plans_to_correct[["id", 'client_id', "name", "lpf", "is_lpf"]]
```
