```python
from odin import PostgresWrapper
```

```python
identity_db = PostgresWrapper(
    instance_connection_name="rmr-cloud-services:us-west3:gateway-services",
    db_name=...,
    user=...,
    password=...,
    ip_type=...,
)
```

```python
identity_db.list_tables()
```

```python

ENTITY_SCHEMA = "public"
ENTITY_TABLE = "entity"


ENTITY_COLUMNS = (
    "task_id",
    "name",
    "status",
    "employer_id",
    "account_manager",
    "cobra_manager",
    "date_admin_start",
    "name_operational",
)

def search_clients_by_rmrcode(
    rmrcode: str,
) -> list[dict]:
    """Partial, case-insensitive match on the client's rmr code."""

    # Escaped so a caller's % or _ matches literally instead of as a wildcard.
    escaped = rmrcode.replace("\\", "\\\\").replace("%", "\\%").replace("_", "\\_")

    rows = identity_db.query(
        f'SELECT {", ".join(ENTITY_COLUMNS)} FROM "{ENTITY_SCHEMA}"."entities"'
        " WHERE employer_id ILIKE %s ESCAPE '\\'"
        " ORDER BY employer_id"
        " LIMIT %s",
        [f"%{escaped}%", 10],
    )

    return rows


results = search_clients_by_rmrcode("RMRUSAD")
```
