```python
'''
    -- get the value of the custom field named "Review Status"

    SELECT jsonb_path_query_first(
        raw_json,
        '$.custom_fields[*] ? (@.name == "Review Status").value'
    ) AS review_status_value
    FROM datafiles;
'''
```

```python
'''
    -- find rows where the FTP User custom field equals a specific value
    
    SELECT * FROM datafiles
    WHERE jsonb_path_exists(
        raw_json,
        '$.custom_fields[*] ? (@.name == "FTP User" && @.value == "client_smith")'
    );
'''
```
