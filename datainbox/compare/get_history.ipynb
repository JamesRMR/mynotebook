```python
def search_clickup_gateway(path, filename, file_date):
    test_row = test_list_df[
        (test_list_df['Task Name'] == filename)
    ]

    if not test_row.empty:
        return test_row.iloc[0]

def search_clickup_datainbox(path, filename, file_date):
    orig_row = orig_list_df[
        (orig_list_df['Task Name'] == filename)
    ]

    if not orig_row.empty:
        return orig_row.iloc[0]

def search_database_gateway(path, filename, file_date):
    pass

def search_database_coredb(path, filename, file_date):
    pass
```

```python
action_dicts = []

for action in action_log.data:

    try:

        path_split = action.path.split("/")
        path = "/".join(path_split[:-1])
        filename = path_split[-1]

    except AttributeError:
        path = ""
        filename = ""


    created_at = action.created_at
    action_type = action.action

    action_dict = {
        "path": path,
        "filename": filename,
        "created_at": created_at,
        "action": action_type,
        "clickup_datainbox_found": None,
        "clickup_gateway_found": None,
        "database_gateway_found": None,
        "database_coredb_found": None
    }

    if action_type in ["login", "read"]:
        continue

    # look for file in the original datainbox list
    orig_row = search_clickup_datainbox(path, filename, created_at)
    if orig_row is not None:
        action_dict['clickup_datainbox_found'] = orig_row['Task ID']

    # look for the file in the gateeway test list
    test_row = search_clickup_gateway(path, filename, created_at)
    if test_row is not None:
        action_dict['clickup_gateway_found'] = test_row['Task ID']


    action_dicts.append(action_dict)
```
